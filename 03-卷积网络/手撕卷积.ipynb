```python
import torch
import torch.nn as nn

def conv_example(in_channel,kernel):
    # in_channel=(28,28)
    # kernel: (5,5)
    output=torch.zeros(24,24)
    for h in range(24):# 行数
        for w in range(24): #列数 开始每5个向外扩展
            inputs =in_channel[h:h+5,w:w+5]
            output[h,w]=(inputs*kernel).sum()# *逐元素相乘
    return output
```

```python
m=nn.Conv2d(1,1,(5,5),bias=False)
x=torch.rand(1,1,28,28)

re=m(x)
re.shape
```

```python
m.weight.shape
```

```python
re1 =conv_example(x.squeeze(0,1),m.weight.squeeze(0,1))
re1.shape
```

```python
torch.all((re-re1).abs()<0.001)
```
